```python
root=Path('outputs')/MODE/TOKEN_SCHEMA/DATASET
metrics=pd.read_csv(root/'metrics_by_seed.csv') if (root/'metrics_by_seed.csv').exists() else pd.DataFrame()
display(metrics.head())
```

```python
if len(metrics):
    fig,ax=plt.subplots(figsize=(5,3.5))
    fixed=metrics.query("section == 'fixed_input_q'").groupby('total_qubits', as_index=False).mean(numeric_only=True)
    ax.plot(fixed.total_qubits, fixed.test_auc, 'o-', label='QRC')
    ax.plot(fixed.total_qubits, fixed.crc3_z_zz_test_auc, 'D--', label='CRC3 z/zz (no PCA)')
    ax.set(xlabel='total Q (fixed injected information)', ylabel='test AUC')
    ax.grid(alpha=.25); ax.legend(frameon=False)
```
