# Graph-stream validity: training, analysis, and visualization

This notebook calls the same saved plans as the CLI. `quick` writes to `outputs_quick/`; it never contaminates confirmatory `outputs/`. Change `PLAN`, `PROFILE`, and `RUN_EXPERIMENT` explicitly.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display, Image

ROOT = Path.cwd().resolve()
if ROOT.name != '5_graph_stream_validity':
    matches = [p for p in [ROOT, *ROOT.parents] if p.name == '5_graph_stream_validity']
    if not matches:
        ROOT = Path('/home/ubuntu/Workspace/Fei/wlf/Article/QRC/FINAL/5_graph_stream_validity')
    else:
        ROOT = matches[0]
sys.path.insert(0, str(ROOT))
from validity_core import run_plan

PLAN = ROOT / 'BACE/01_isomorphism_robustness/Q16_T16/plan.json'
PROFILE = 'quick'  # quick or full
RUN_EXPERIMENT = False  # set True deliberately
```

```python
# Training/feature extraction. QRC seeds are split across the plan's GPU devices.
if RUN_EXPERIMENT:
    result = run_plan(PLAN, profile_override=PROFILE)
    display(result)
else:
    print('Dry notebook: set RUN_EXPERIMENT=True to execute the selected plan.')
```

```python
# Analysis: load the seed-level and summary evidence generated by any plan.
output_name = 'outputs_quick' if PROFILE == 'quick' else 'outputs'
OUT = PLAN.parent / output_name
for name in [
    'frozen_cross_variant_summary.csv',
    'frozen_cross_variant_effects.csv',
    'representation_robustness.csv',
    'representation_order_sensitivity.csv',
    'collision_label_conflict_Q_T.csv',
]:
    path = OUT / name
    if path.exists():
        print(name)
        display(pd.read_csv(path).head(20))
```

```python
# Visualization: plans save PNG/PDF/SVG from the same source CSVs.
for figure in sorted((OUT / 'figures').glob('*.png')):
    print(figure.name)
    display(Image(filename=str(figure)))
```
